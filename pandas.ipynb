```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({
    'Name': ['Nikita Rai', 'Alisha Sharma', 'Nishan Kumar'],
    'City': ['Kathmandu', 'Pokhara', 'Kathmandu'],
    'Age': [23, 25, 26],
    'Salary': [50000, 70000, 60000],
    'Date': pd.to_datetime(['2026-01-10', '2026-02-15', '2026-03-20'])
})

# Feature Engineering
df['First_Name'] = df['Name'].apply(lambda x: x.split()[0])
df['Salary_Level'] = df['Salary'].apply(lambda x: 'High' if x > 60000 else 'Low')
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month

# Encoding
le = LabelEncoder()
df['City'] = le.fit_transform(df['City'])

# Drop original column
df.drop(['Name', 'Date'], axis=1, inplace=True)

print(df)
```
