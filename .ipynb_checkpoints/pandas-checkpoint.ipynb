```python
# Pandas Series
# A Pandas Series is one of the most important data structures in the Python library pandas.
# It is basically a 1-dimensional labeled array, similar to a column in a table.
```

```python
# creating series 
import numpy as np
import pandas as pd
```

```python
labels = ['a','b','c','d']
my_list = [10,20,30,40]
arr = np.array([10,20,30])
d = {1:10,2:20,3:30} 
pd.Series(my_list)
```

```python
pd.Series(my_list,index=labels)
```

```python
pd.Series(arr)
```

```python
pd.Series(d)
```

```python
# Data Frame
# creating dataFrame

data = {
    'Name': ['Nikita', 'Alisha', 'Gita', 'Nishan'],
    'Age': [23, 44, 23, 23],
    'City': ['Kathmandu', 'Pokhara', 'Chitwan', 'Sindhuli'],
    'Salary': [50000, 70000, 45000, 60000]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
data_list = [
    ['Nikita', 23, 'Kathmandu', 50000],
    ['Alisha', 44, 'Pokhara', 70000],
    ['Gita', 23, 'Chitwan', 45000],
    ['Nishan', 23, 'Sindhuli', 60000]
]
df2 = pd.DataFrame(data_list)
df2
```

```python
columns=["Name", "Age", "City", "Salary"]
df2 = pd.DataFrame(data_list, columns=columns)
df2
```
