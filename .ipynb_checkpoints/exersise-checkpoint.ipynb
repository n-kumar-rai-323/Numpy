```python
import numpy as np
```

```python
s = np.array([
    [5,3,4,6,7,8,9,1,2],
    [6,7,2,1,9,5,3,4,6],
    [1,9,8,3,4,2,5,6,7],

    [2,7,6,1,5,9,4,3,8],
    [9,5,1,4,3,8,2,7,6],
    [4,8,3,7,2,6,1,5,9],

    [8,1,2,7,5,3,6,4,9],
    [6,5,7,1,9,4,8,2,3],
    [4,9,3,6,8,2,7,1,5]
])

# Row check
for row in s:
    if np.sum(row) != 45:
        print("Sudoku is not valid (row issue)")
        break
else:
    print("Rows are valid")

# Column check (FIXED)
for col in s.T:   # or axis=0
    if np.sum(col) != 45:
        print("Sudoku is not valid (column issue)")
        break
else:
    print("Columns are valid")
```
