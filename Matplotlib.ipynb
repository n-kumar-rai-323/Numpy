```python
import numpy as np
x = np.linspace(0,5,11)
x
y = x ** 2
y
```

```python
import matplotlib.pyplot as plt
plt.plot(x,y)
```

```python
plt.plot(x,y)
plt.title("Matplotlib")
plt.xlabel("x plot")
plt.ylabel("y plot")
```

```python
# subplots in matplotib
plt.subplot(2,2,1)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
```

```python
# object oriented way
fig= plt.figure()
fig.add_axes([0.1,0.1,0.8,0.8]) 
```

```python
fig= plt.figure()
axis1 = fig.add_axes([0.1,0.1,1,1]) 
axis1.plot(x,y)
axis2 = fig.add_axes([0.4,0.5,0.3,0.3])
axis2.plot(x,y)
```
