```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ram", "Sita", "Hari", "Gita", "Mohan", "Rita"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [50000, 45000, 55000, 70000, 48000, 60000],
    "Age": [25, 30, 28, 35, 32, 27]
})
```

```python
df.groupby("Department")["Salary"].mean()
```

```python
# Sum Salary
df.groupby("Department")["Salary"].sum()
```

```python
# Count Employees
df.groupby("Department")["Name"].count()
```
