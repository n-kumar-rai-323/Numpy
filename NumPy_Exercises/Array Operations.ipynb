```python
import numpy as np
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

print("Addition: ", a+b)
print("Subtraction: ", a -b)
print("Multiplication: ", a * b)
print("Division: ",a /b)
print("Square root", np.sqrt(a))
print("Find Maximum: ",a.max())
print("Find Minimum: ", a.min())
print("Find Sum: ", a.sum())
print("Find Mean:", a.mean())
```
